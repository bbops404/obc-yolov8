```python

# Jupyter Notebook to load YOLOv8-CA pretrained FP32 weights and save layer tensors

import torch
from ultralytics import YOLO

# Path to your trained FP32 weights
weights_path = '/Users/user/Documents/obc-yolov8/obc-yolov8/runs/detect/train3/weights/last.pt'

# Load model with pretrained weights
model = YOLO('/Users/user/Documents/obc-yolov8/ultralytics10.24/ultralytics/cfg/models/v8/yolov8-CA.yaml')
model.load(weights_path)

# Move to CPU
model.to('cpu')

# Function to save FP32 weights of all layers
for i, layer in enumerate(model.model): # type: ignore
    if hasattr(layer, 'weight'):
        weights = layer.weight.data # type: ignore
        filename = f'layer{i}_weights_fp32.pt'
        torch.save(weights, filename)
        print(f'Saved FP32 weights of layer {i} -> {filename}')

# For the Detect head
detect_layer = model.model[-1] # type: ignore
for i, conv in enumerate(detect_layer.m):
    weights = conv.weight.data
    filename = f'detect_head_conv{i}_weights_fp32.pt'
    torch.save(weights, filename)
    print(f'Saved FP32 weights of detect head conv {i} -> {filename}')
```
